```python
# Create a simple custom tool that returns today's date using Python's datetime function.
from datetime import datetime
from agno.agent import Agent

def get_today_date() -> str:
    """
    Returns today's date
    """
    today = datetime.now().strftime("Today is %B %d %Y")
    return today

get_today_date()
```

#### Register the custom date function as a tool that the Groq-powered Agno Agent can use.

```python
from agno.models.groq import Groq

agent = Agent(
    model=Groq(id="openai/gpt-oss-120b"),
    tools=[get_today_date],
    markdown=True
)
```

#### Call the agent and let it use the custom date tool to answer the user's question.

```python
# Call the custom date tool and display the current date
print(get_today_date())
```

```python
# Call the agent and let it use the custom date tool to answer the user's question.
agent.print_response("What is today's date?", stream=True)
```

#### Create a custom Wikipedia summarizer tool that fetches and returns a topic summary using Wikipedia's REST API.

```python
import httpx
from agno.agent import Agent

def search_wikipedia(topic: str = "Machine Learning") -> str:
    url = f"https://en.wikipedia.org/api/rest_v1/page/summary/{topic.replace(' ', '%20')}"

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    response = httpx.get(url, headers=headers, timeout=5.0)
    data = response.json()

    # Extract the summary from response
    if "extract" in data:
        title = data.get("title", topic)
        extract = data["extract"]
        return f"**{title}**:\n{extract}"
    else:
        return "Sorry, I couldn't find anything on that topic."
```

```python
from agno.models.groq import Groq

agent = Agent(
    model=Groq(id="openai/gpt-oss-120b"),
    tools=[search_wikipedia],
    markdown=True
)
```

```python
agent.print_response("Explain quantum computing in simple words",stream=True)
```

#### - Now we will explore the concept of Memory in Agentic AI.
#### - Memory allows an AI Agent to remember previous conversations and maintain context about the user.
#### - This helps the Agent understand who it is talking to, the user's name and preferences,
#### - and what was discussed in previous conversations.
#
#### In Agno, we will explore three important types of memory:
1. Session Storage  → Stores the data and conversation history of a particular session.
2. User Memory      → Stores important information about the user across conversations.
3. Session Summaries → Creates a concise summary of previous conversations to preserve context.
#
#### - Ab hum Agentic AI mein Memory ka concept samjhenge.
#### - Memory ki madad se AI Agent previous conversations ko yaad rakh sakta hai
#### - aur user ke saath conversation ka context maintain kar sakta hai.
#### - Iska fayda ye hai ki Agent ko pata reh sakta hai ki woh kis user se baat kar raha hai,
#### - user ka naam kya hai, uski preferences kya hain, aur past mein kya discussion hui thi.
#
#### - Agno mein hum Memory ke teen important types explore karenge:
1. Session Storage      → Ek particular session ka data aur conversation history store karta hai.
2. User Memory          → User ki important information ko different conversations mein remember karta hai.
3. Session Summaries    → Previous conversation ka short summary banakar context ko preserve karta hai.

```python
from agno.agent import Agent
from agno.memory.v2.memory import Memory
from agno.memory.v2.db.sqlite import SqliteMemoryDb
from agno.storage.sqlite import SqliteStorage
from rich.pretty import pprint
```

#### Create a Groq-powered AI Agent with persistent user memory, session storage, and conversation history.

```python
from agno.models.groq import Groq

db_file = "temp/prajwal_agent_storage.db"

user_id = "prajwal_id"

memory = Memory(
    db=SqliteMemoryDb(
        table_name="user_memories",
        db_file=db_file
    ),
)

agent = Agent(
    model=Groq(id="openai/gpt-oss-120b"),
    description="You're an AI with a memory!",
    memory=memory,
    storage=SqliteStorage(
        table_name="agent_sessions",
        db_file=db_file
    ),
    enable_user_memories=True,
    add_history_to_messages=True,
    num_history_runs=3,
    session_id="prajwal_chat_session",
    markdown=True,
)
```

#### Test the agent's ability to remember user preferences across conversations.a

```python
agent.print_response(
    "I really enjoy playing cricket and watching cricket matches. What sport do you like the most?",
    user_id=user_id
)
```

```python
print("\n Updated **memories** about the user:")
user_memories = memory.get_user_memories(user_id=user_id)
pprint([{"memory": m.memory, "topics": m.topics} for m in user_memories])
```

#### Retrieve and display the user's stored memories to verify what the Agent remembers from previous conversations.
